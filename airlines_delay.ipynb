```python
# Function to detect outliers using IQR
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers
```

```python
for i in df1:
    if df1[i].dtypes in [int,float]:
        outliers = detect_outliers_iqr(df1, i)
        print(f"Feature: {i} → Outliers detected: {len(outliers)}")
```

```python
def remove_outliers(df, col):
    Q1 = df[col].quantile(0.25)  # 25th percentile
    Q3 = df[col].quantile(0.75)  # 75th percentile
    IQR = Q3 - Q1  # Interquartile range
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
```

```python
while True:
    prev_shape = df1.shape[0]
    for column in df1.select_dtypes(include=['int64', 'float64']).columns:
        df1 = remove_outliers(df1, column)
    # Stop when no more rows are dropped
    if df1.shape[0] == prev_shape:
        break
```

```python
df=pd.get_dummies(df1,drop_first=True,columns=['Airline','AirportFrom','AirportTo'])
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
```

```python
pd.set_option('display.max_columns', None)
df[(df['id']==5)]
```

```python
X=df.drop(columns=['Delay','id','Flight'])
y = df['Delay']
X_scaled=scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=200)
```

Classification

```python
import xgboost as xgb
```

```python
model=xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Report:\n", classification_report(y_test, y_pred))
```
